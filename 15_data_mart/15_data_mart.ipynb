## 1. Why data marts exist

A mart is a curated analytical subset/aggregate for a business area or reporting use case.

## 2. Product sales mart

Create an aggregate that BI can consume without repeating heavy logic.

```
import duckdb
con=duckdb.connect('../14_data_warehouse/data/icecream_dw.duckdb')
product_mart=con.sql("SELECT p.category,p.product_name,SUM(f.quantity) units,SUM(f.sales_amount) sales,SUM(f.profit) profit FROM fact_sales f JOIN dim_product p ON f.product_key=p.product_key GROUP BY 1,2").df()
product_mart.to_csv('data/dm_product_sales.csv',index=False)
product_mart.head()
```

## 3. Store and monthly marts

Build multiple business-ready marts.

```
store=con.sql("SELECT s.store_name,SUM(f.sales_amount) sales,SUM(f.profit) profit FROM fact_sales f JOIN dim_store s ON f.store_key=s.store_key GROUP BY 1").df()
store.to_csv('data/dm_store_sales.csv',index=False)
month=con.sql("SELECT d.year,d.month,d.month_name,SUM(f.sales_amount) sales,SUM(f.profit) profit FROM fact_sales f JOIN dim_date d ON f.date_id=d.date_id GROUP BY 1,2,3 ORDER BY 1,2").df()
month.to_csv('data/dm_monthly_sales.csv',index=False)
```
