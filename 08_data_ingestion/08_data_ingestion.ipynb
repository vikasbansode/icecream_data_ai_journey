# 08 Data Ingestion — Source to Raw

This notebook is a practical, executable lesson. Each step follows **Context → Source Problem → Detect → Fix → Verify → Business Result → Exercise**.

## 1. Ingestion principle

Ingestion moves collected source data into the raw layer. Raw should be reproducible and close to source.

## 2. Create raw folders

Separate source systems so students can see where each raw object came from.

```
from pathlib import Path
raw=Path('data/raw')
for x in ['pos','customer','product','store','weather','supplier','competitor','sensor','reference']:
    (raw/x).mkdir(parents=True,exist_ok=True)
print('raw folders created')
```

## 3. Ingest source files

Copy collected files without applying business transformations.

```
import shutil
from pathlib import Path
files={
 '../07_data_collection/data/collected_sales_transactions.csv':'data/raw/pos/sales_transactions.csv',
 '../07_data_collection/data/collected_customer_registry.csv':'data/raw/customer/customer_registry.csv',
 '../07_data_collection/data/collected_product_catalog.csv':'data/raw/product/product_catalog.csv',
 '../07_data_collection/data/collected_weather.csv':'data/raw/weather/weather_station_daily.csv'}
for a,b in files.items(): shutil.copy2(a,b)
print('ingested',len(files),'source objects')
```

## 4. Raw validation

Validate that ingestion did not silently change row counts.

```
import pandas as pd
raw_sales=pd.read_csv('data/raw/pos/sales_transactions.csv')
source_sales=pd.read_csv('../07_data_collection/data/collected_sales_transactions.csv')
assert len(raw_sales)==len(source_sales)
print('row-count check passed:',len(raw_sales))
```
