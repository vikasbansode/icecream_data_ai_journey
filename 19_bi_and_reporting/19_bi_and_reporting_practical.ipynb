## 1. KPI definitions

Define each KPI with formula, grain, filters and owner before building a dashboard.

```
kpis={
'Total Sales':'SUM(sales_amount)',
'Total Units':'SUM(quantity)',
'Transactions':'COUNT(DISTINCT transaction_id)',
'Profit':'SUM(profit)',
'Average Order Value':'SUM(sales_amount)/COUNT(DISTINCT transaction_id)'
}
for k,v in kpis.items(): print(k,'=',v)
```

## 2. Create reporting dataset

Use a mart as the stable BI input.

```
import pandas as pd
report=pd.read_csv('../15_data_mart/data/dm_monthly_sales.csv')
report.head()
```

## 3. Create Excel-style report

Export a management-ready dataset.

```
report.to_excel('data/monthly_management_report.xlsx',index=False)
print('report written')
```

## 4. Dashboard-ready Plotly

Create a small interactive report view.

```
import plotly.express as px
px.bar(report,x='month_name',y='sales',title='Monthly Sales').show()
```
