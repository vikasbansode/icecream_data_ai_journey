## 1. Raw layer rule

Do not clean, standardize or model raw data. Preserve source values so the pipeline can be replayed.

## 2. Inspect raw POS

Use pandas to inspect raw data without modifying it.

```
import pandas as pd
sales=pd.read_csv('../08_data_ingestion/data/raw/pos/sales_transactions.csv')
print(sales.head())
print(sales.dtypes)
```

## 3. Raw profiling with Polars

Polars is useful for fast profiling and transformations on tabular data.

```
import polars as pl
df=pl.read_csv('../08_data_ingestion/data/raw/pos/sales_transactions.csv')
print(df.shape)
print(df.null_count())
```

## 4. Raw querying with DuckDB

DuckDB can query files directly, which is useful before loading a warehouse.

```
import duckdb
print(duckdb.sql("SELECT store_id, SUM(sales_amount) AS sales FROM read_csv_auto('../08_data_ingestion/data/raw/pos/sales_transactions.csv') GROUP BY store_id ORDER BY sales DESC"))
```
