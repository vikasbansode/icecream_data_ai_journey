## 1. Scope

This phase stops at business analysis and BI. No machine learning, deep learning, GenAI, RAG or agents are used here.

## 2. Descriptive — What / How much / How many

Use DuckDB and pandas to answer sales, transaction, units and profit questions.

```
import duckdb,pandas as pd
con=duckdb.connect('../14_data_warehouse/data/icecream_dw.duckdb')
summary=con.sql("SELECT SUM(sales_amount) sales,SUM(quantity) units,COUNT(*) transactions,SUM(profit) profit FROM fact_sales").df()
summary
```

## 3. When / Where / Which / Who

Drill down by date, store, product and customer.

```
q=con.sql("SELECT s.store_name,SUM(f.sales_amount) sales FROM fact_sales f JOIN dim_store s ON f.store_key=s.store_key GROUP BY 1 ORDER BY 2 DESC")
q.df()
```

## 4. Diagnostic — Why?

Compare performance across dimensions and investigate variance rather than predicting the future.

```
monthly=con.sql("SELECT d.year,d.month_name,SUM(f.sales_amount) sales FROM fact_sales f JOIN dim_date d ON f.date_id=d.date_id GROUP BY 1,2 ORDER BY 1,MIN(d.month)").df()
monthly
```

## 5. Pandas analysis

Pandas supports flexible ad-hoc analysis.

```
df=con.sql('SELECT * FROM fact_sales').df()
print(df.describe())
```

## 6. Polars analysis

Polars provides an alternative dataframe workflow.

```
import polars as pl
pl_df=pl.from_pandas(df)
print(pl_df.group_by('product_key').agg(pl.col('sales_amount').sum().alias('sales')).sort('sales',descending=True).head())
```
