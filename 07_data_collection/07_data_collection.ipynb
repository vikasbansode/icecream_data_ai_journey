# 07 Data Collection — Collect Source Data as It Exists

This notebook is a practical, executable lesson. Each step follows **Context → Source Problem → Detect → Fix → Verify → Business Result → Exercise**.

## 1. Collection vs discovery

Discovery describes the source; collection obtains a copy/snapshot/response. Collection must preserve provenance.

## 2. Collect SQLite sales

Copy/extract the source table into a collected CSV while recording source metadata.

```
from pathlib import Path
import sqlite3,pandas as pd,json,datetime,shutil
Path('data').mkdir(exist_ok=True)
con=sqlite3.connect('../05_data_sources/source_data/sqlite/sales_source.db')
df=pd.read_sql_query('SELECT * FROM sales_transactions',con)
df.to_csv('data/collected_sales_transactions.csv',index=False)
metadata={'source':'sales_source.db','table':'sales_transactions','collected_at':datetime.datetime.now().isoformat(),'rows':len(df)}
Path('data/collection_metadata.json').write_text(json.dumps(metadata,indent=2))
print(metadata)
```

## 3. Collect CSV, Excel, XML, API and sensor sources

Each source is collected with its original meaning. Do not create dimensions here.

```
import pandas as pd, json, xml.etree.ElementTree as ET
customer=pd.read_csv('../05_data_sources/source_data/csv/customer_registry.csv')
product=pd.read_excel('../05_data_sources/source_data/excel/product_catalog.xlsx')
weather=pd.read_csv('../05_data_sources/source_data/csv/weather_station_daily.csv')
customer.to_csv('data/collected_customer_registry.csv',index=False)
product.to_csv('data/collected_product_catalog.csv',index=False)
weather.to_csv('data/collected_weather.csv',index=False)
print(customer.shape,product.shape,weather.shape)
```

## 4. Collection log

Record source, extraction time, row count and file name. This becomes lineage evidence.

```
from pathlib import Path
import json,datetime
log={'collected_at':datetime.datetime.now().isoformat(),'artifacts':['collected_sales_transactions.csv','collected_customer_registry.csv','collected_product_catalog.csv','collected_weather.csv']}
Path('data/collection_log.json').write_text(json.dumps(log,indent=2))
print(log)
```
