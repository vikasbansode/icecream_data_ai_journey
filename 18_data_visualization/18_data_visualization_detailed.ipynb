## 1. Visualization principle

Choose a chart because it answers a business question, not because it looks attractive.

## 2. Sales by product

Use Plotly for interactive business visuals.

```
import duckdb,plotly.express as px
con=duckdb.connect('../14_data_warehouse/data/icecream_dw.duckdb')
df=con.sql("SELECT p.product_name,SUM(f.sales_amount) sales FROM fact_sales f JOIN dim_product p ON f.product_key=p.product_key GROUP BY 1 ORDER BY 2 DESC").df()
fig=px.bar(df,x='product_name',y='sales',title='Sales by Product')
fig.show()
```

## 3. Sales trend

Line charts answer WHEN questions.

```
df=con.sql("SELECT d.date,SUM(f.sales_amount) sales FROM fact_sales f JOIN dim_date d ON f.date_id=d.date_id GROUP BY 1 ORDER BY 1").df()
px.line(df,x='date',y='sales',title='Daily Sales Trend').show()
```

## 4. Store comparison

Bar charts help compare stores.

```
df=con.sql("SELECT s.store_name,SUM(f.sales_amount) sales FROM fact_sales f JOIN dim_store s ON f.store_key=s.store_key GROUP BY 1 ORDER BY 2 DESC").df()
px.bar(df,x='store_name',y='sales',title='Store Sales').show()
```
