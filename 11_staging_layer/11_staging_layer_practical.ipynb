## 1. Why staging exists

Staging is where cleaned, standardized source data is shaped consistently for transformation and modeling.

## 2. Load cleaned source objects

Build standardized staging datasets without yet creating dimensional warehouse tables.

```
import pandas as pd
from pathlib import Path
stage=Path('data')
stage.mkdir(exist_ok=True)
sales=pd.read_csv('../10_data_cleaning/data/clean_sales.csv')
customers=pd.read_csv('../07_data_collection/data/collected_customer_registry.csv')
products=pd.read_csv('../07_data_collection/data/collected_product_catalog.csv')
weather=pd.read_csv('../07_data_collection/data/collected_weather.csv')
print(sales.shape,customers.shape,products.shape,weather.shape)
```

## 3. Standard staging names and types

Use consistent column names and data types across source systems.

```
sales.columns=[c.lower() for c in sales.columns]
sales['date_id']=sales['date_id'].astype(str) if 'date_id' in sales else sales.get('date_id')
sales.to_csv('data/stg_sales.csv',index=False)
customers.columns=[c.lower().strip() for c in customers.columns]
customers.to_csv('data/stg_customer.csv',index=False)
products.columns=[c.lower().strip() for c in products.columns]
products.to_csv('data/stg_product.csv',index=False)
```

## 4. Staging validation

Check keys and referential readiness before dimensional modeling.

```
print('customer ids:',customers['customer_id'].nunique())
print('product ids:',products['product_id'].nunique())
print('sales customer ids missing from customer source:',len(set(sales.customer_id)-set(customers.customer_id)))
print('sales product ids missing from product source:',len(set(sales.product_id)-set(products.product_id)))
```
