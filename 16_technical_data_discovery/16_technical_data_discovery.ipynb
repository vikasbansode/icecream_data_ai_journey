Now that a database exists, database metadata discovery becomes meaningful. This is where information_schema/system catalogs belong.

## 2. DuckDB information_schema

Discover tables and columns in the analytical database.

```
import duckdb
con=duckdb.connect('../14_data_warehouse/data/icecream_dw.duckdb')
print(con.sql("SELECT table_name,column_name,data_type FROM information_schema.columns ORDER BY table_name,ordinal_position"))
```

## 3. PostgreSQL information_schema example

Show the equivalent query students can use when the same model is deployed to PostgreSQL.

```
sql="""SELECT table_schema, table_name, column_name, data_type
FROM information_schema.columns
WHERE table_schema NOT IN ('pg_catalog','information_schema')
ORDER BY table_schema,table_name,ordinal_position;"""
print(sql)
```
