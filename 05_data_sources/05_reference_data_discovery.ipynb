# Reference Data Discovery — Frosty World

Reference data provides controlled, standardized values used to classify and validate data.

## 1. Examples
Customer Segment, Product Category, Payment Method, Sales Channel, Order Status, Weather Condition, UOM, Tax Category, Store Type and Geography.

```
from pathlib import Path
import pandas as pd
base=Path('../05_data_sources/reference_data')
segment=pd.read_csv(base/'ref_customer_segment.csv')
category=pd.read_csv(base/'ref_product_category.csv')
weather=pd.read_csv(base/'ref_weather_condition.csv')
segment
```

## 2. Allowed-value validation
A transaction or master record should use an allowed reference value.

```
allowed=set(segment['customer_segment_name'])
sample=pd.DataFrame({'segment':['Family','student','Unknown','Corporate']})
sample['valid']=sample['segment'].isin(allowed)
sample
```

## 3. Standardization
Source systems may use different representations such as student, STUDENT or Student. Mapping rules standardize them to one governed value.

```
mapping=pd.read_csv(base/'reference_mapping_examples.csv')
mapping
```

## 4. Geography hierarchy
Country -> State -> City is a reference hierarchy used consistently across customer, store and weather data.

```
geo=pd.read_csv(base/'ref_geography.csv')
geo
```

## 5. Governance questions
Who owns the reference domain? Who approves a new value? What is the effective date? Can a value be retired? How are old source values mapped?
