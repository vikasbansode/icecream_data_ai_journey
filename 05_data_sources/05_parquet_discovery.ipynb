# Source Data Discovery — Parquet

This is a complete source-specific discovery lesson. Discovery happens **before or during collection/ingestion** so that students understand what actually exists at the source.

## Discovery checklist

- File Metadata
- Schema
- Columns
- Logical Types
- Statistics
- Compression
- Partitioning
- Row Groups
- Null Counts
- Predicate Pushdown Concept

## Why this matters

A source is not just a file or endpoint. We need to understand its structure, business meaning, grain, quality, technical limitations, ownership and change behavior before designing ingestion.

## Student discovery record

Students must document: source name, source owner, location/endpoint, structure, entities, attributes, data types, grain, keys, nullability, sample values, frequency, history, quality risks, security/privacy considerations, expected change patterns and recommended collection method.

## Exercises

1. Profile the supplied source.
2. Identify all entities and attributes.
3. Identify candidate keys.
4. Determine the grain.
5. Identify at least five data-quality risks.
6. Recommend a collection/ingestion approach.
7. Write a source-to-business mapping.
8. Explain what you would preserve in the raw layer.
