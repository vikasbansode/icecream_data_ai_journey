# Master Data Discovery — Frosty World

This notebook teaches Master Data as business-entity data that is shared across processes. It is intentionally detailed and practical.

## 1. What is Master Data?
Master data describes stable business entities such as Customer, Product, Store, Supplier, Employee and Promotion. It is different from transaction data such as a sale.

## 2. Why Master Data Matters
If two systems use different product names, customer identifiers or store codes, joins and reports become unreliable. Master data provides a common business identity.

## 3. Master Data in Frosty World
- Customer Master
- Product Master
- Store Master
- Supplier Master
- Employee Master
- Promotion Master

```
from pathlib import Path
import pandas as pd
base = Path('../05_data_sources/master_data')
customer = pd.read_csv(base/'customer_master.csv')
product = pd.read_csv(base/'product_master.csv')
store = pd.read_csv(base/'store_master.csv')
customer.head()
```

## 4. Discover the structure
Inspect row counts, columns, data types, nulls, duplicates and candidate keys.

```
for name, df in {'customer':customer,'product':product,'store':store}.items():
    print('---', name, '---')
    print('rows:', len(df))
    print(df.dtypes)
    print('duplicates:', df.duplicated().sum())
    print('nulls:')
    print(df.isna().sum())
```

## 5. Candidate business keys
Customer ID, Product ID and Store ID should be unique within the mastered entity.

```
print('Customer duplicate IDs:', customer['customer_id'].duplicated().sum())
print('Product duplicate IDs:', product['product_id'].duplicated().sum())
print('Store duplicate IDs:', store['store_id'].duplicated().sum())
```

## 6. Lifecycle and governance attributes
Master records include status, effective dates and source system so students can discuss ownership, lifecycle and history.

## 7. Master Data vs Dimension
A Product Master is an operational mastered entity. A Product Dimension is an analytical structure. The dimension may contain surrogate keys and SCD history.
