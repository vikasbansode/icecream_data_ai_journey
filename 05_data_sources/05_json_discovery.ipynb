# Source Data Discovery — JSON

This is a complete source-specific discovery lesson. Discovery happens **before or during collection/ingestion** so that students understand what actually exists at the source.

## Discovery checklist

- Top-Level Object/Array
- Keys
- Nested Objects
- Arrays
- Optional Fields
- Data Types
- Schema Inference
- Nulls
- Repeated Structures
- Flattening
- Grain

## Why this matters

A source is not just a file or endpoint. We need to understand its structure, business meaning, grain, quality, technical limitations, ownership and change behavior before designing ingestion.

## Interpretation and business decision

Do not end the lesson at code output. Students must answer:

- What did we learn?
- How reliable is the result?
- What evidence supports the conclusion?
- What evidence is missing?
- What should management do?
- What should be monitored next?

This is the standard used throughout the journey: **data result → insight → decision → action → measurement**.
